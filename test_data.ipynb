```python
import pandas as pd

df = pd.read_excel(
    "DATA DASHBOARD HRD PT ALDZAMA.xlsx",
    sheet_name="INTERNSHIP",
    skiprows=2,     # karena mulai dari baris 3
    usecols="B:L"   # sesuaikan jumlah kolom kamu
)

df.columns = [
    "nama", "program", "institusi", "kota", "ket",
    "permohonan", "penempatan", "durasi",
    "status", "final_project", "potensi"
]

df.head()
```

```python
# ubah permohonan jadi datetime
df['permohonan'] = pd.to_datetime(df['permohonan'], errors='coerce')

# ambil bulan
df['bulan'] = df['permohonan'].dt.to_period('M').astype(str)
```

```python
# 1. jumlah per institusi
institusi = df['institusi'].value_counts().reset_index()
institusi.columns = ['institusi', 'jumlah']

# 2. permohonan per bulan
permohonan = df['bulan'].value_counts().sort_index().reset_index()
permohonan.columns = ['bulan', 'jumlah']

# 3. status intern
status = df['status'].value_counts().reset_index()
status.columns = ['status', 'jumlah']
```

```python
df_pay = pd.read_excel(
    FILE_PATH,
    sheet_name="Pay MCU 2026",
    usecols="T:W"
)

df_pay.columns = [
    "user",
    "project",
    "periode",
    "pembayaran"
]

df_pay.head()
```

```python
# CLEAN STRING
df_pay['user'] = df_pay['user'].astype(str).str.strip()
df_pay['project'] = df_pay['project'].astype(str).str.strip()
df_pay['periode'] = df_pay['periode'].astype(str).str.strip()

# CLEAN UANG (Rp)
df_pay['pembayaran'] = (
    df_pay['pembayaran']
    .astype(str)
    .str.replace('Rp', '')
    .str.replace('.', '', regex=False)
)

df_pay['pembayaran'] = pd.to_numeric(df_pay['pembayaran'], errors='coerce')

# DROP YANG TIDAK ADA NILAI
df_pay = df_pay.dropna(subset=['pembayaran'])

df_pay.head()
```
